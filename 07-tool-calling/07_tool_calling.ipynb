### Connect to Gemini

```python
from google import genai
from google.genai import types
from google.colab import userdata

api_key = userdata.get("GEMINI_API_KEY")
client = genai.Client(api_key=api_key)

MODEL = "gemini-2.5-flash"
```

# Build a mini AI calculator


### Create first tool

```python
def multiply(a: int, b: int) -> int:
    """Multiply two numbers and return the result."""
    result = a * b
    print(f"Tool running: {a} × {b} = {result}")
    return result
```

```python
def add(a:int, b:int)->int:
  """Add two numbers and return the result"""
  result= a+b
  print(f"Tool running: {a}+{b} = {result}")
  return result
```

```python
def subtract(a:int, b:int)->int:
  """Subtract two numbers and return the result"""
  result= a-b
  print(f"Tool running: {a}-{b} = {result}")
  return result
```

```python
def get_square(a:int)->int:
  """Square the number and return the result"""
  result = a*a
  print(f"Square is: {result}")
  return result
```

```python
multiply(10, 5)
```

### Give the tool to Gemini

```python
question = input("Ask your AI calculator: ")

response = client.models.generate_content(
    model=MODEL,
    contents=question,
    config=types.GenerateContentConfig(
        tools=[add, multiply, subtract,  get_square]
    )
)

print("\nAI Answer:")
print(response.text)
```
