### Load an embedding model

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

print("Embedding model loaded!")
```

### Convert text into vectors

```python
sentences = [
    "I love learning Python.",
    "Python programming is interesting to me.",
    "I enjoy eating pizza."
]

embeddings = model.encode(sentences)
print("Shape:", embeddings.shape)
print("First sentence vector:")
print(embeddings[0])
```

### Calculate semantic similarity

```python
from sentence_transformers import util

similarities = util.cos_sim(embeddings, embeddings)

print(similarities.round(decimals=3))
```

### Build a tiny semantic search engine

```python
documents = [
    "Python is a programming language.",
    "Machine learning helps computers learn patterns.",
    "Pizza is a popular Italian food.",
    "Neural networks are used in deep learning.",
    "Python is widely used for AI development."
]

document_embeddings = model.encode(documents)

query = "How can I use Python to build AI applications?"

query_embedding = model.encode(query)

scores = util.cos_sim(
    query_embedding,
    document_embeddings
)[0]

results = sorted(
    zip(documents, scores.tolist()),
    key=lambda item: item[1],
    reverse=True
)

for document, score in results[:3]:
    print(f"{score:.3f} | {document}")
```
