### Using Pydantic

````python
class AIProject(BaseModel):
  name:str
  description:str
  technologies:list[str]
````

````python
project = AIProject(name="RAG Chatbot",description="Chat with documents",technologies=[
        "Python",
        "LLM",
        "Vector Database"
    ])

print(project)
````

````python
print(project.name)
print(project.description)
print(project.technologies)
````

````python
project = AIProject(name="AI Agent",description="Agent that uses tools",technologies=["Python","LLMs","Tools"])
````

````python
print(project.name)
print(project.description)
print(project.technologies)

print(project.model_dump())
````

7. Async API request

````python
import aiohttp
````

````python
async def get_github_user(username):
    url = f"https://api.github.com/users/{username}"

    async with aiohttp.ClientSession() as session:
        async with session.get(url) as response:
            data = await response.json()

            return data
````

````python
user = await get_github_user("fatimarazzaq")

print(user.get("login"))
print(user.get("public_repos"))
````

###Combine everything

````python
class UserProfile(BaseModel):
    username: str
    public_repos: int
    profile_url: str

github_data = await get_github_user("fatimarazzaq")

profile = UserProfile(
    username=github_data.get("login"),
    public_repos=github_data.get("public_repos"),
    profile_url=github_data.get("html_url")
)

print(profile)
````

````python
print(profile.model_dump())
````

**Mini project — AI Project Analyzer**

````python
projects = [
    {
        "name": "RAG Chatbot",
        "description": "Chat with PDF documents",
        "technologies": ["Python", "LLM", "RAG"]
    },
    {
        "name": "AI Agent",
        "description": "Agent that uses external tools",
        "technologies": ["Python", "LLM", "Tools"]
    }
]
````

````python
class Project(BaseModel):
    name: str
    description: str
    technologies: list[str]
````

````python
validated_projects = [Project(**project) for project in projects]
validated_projects
````

````python
for project in validated_projects:
    print(project.name)
    print(project.technologies)
    print()
````
