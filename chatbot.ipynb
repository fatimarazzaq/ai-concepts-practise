3. Create Gemini client

```
SYSTEM INSTRUCTION
"You are a helpful AI tutor..."
          ↓
       Gemini
          ↑
USER INPUT
"What is an API?"
```




````python
from google import genai

client = genai.Client(api_key=api_key)
````

# Step 1 — Add a system instruction

````python
from google.genai import types

response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents="What is an API?",
    config=types.GenerateContentConfig(
        system_instruction="""
        You are a helpful AI tutor.
        Explain technical concepts in simple English.
        Always give a small example.
        """
    )
)

print(response.text)
````

# Step 2 — Create a conversation

````python
chat = client.chats.create(
    model="gemini-3.5-flash"
)
````

````python
response = chat.send_message(message="What is an API?")
print(response.text)
````

````python
response = chat.send_message(
    message="Give me a Python example"
)

print(response.text)
````

````python
response = chat.send_message(
    message="Explain that example line by line"
)

print(response.text)
````

# Step 3 — See the history

````python
history = chat.get_history()

for message in history:
    print(message.role)
    print(message.parts[0].text)
    print("-" * 50)
````

# Step 4 — Making it interactive

````python
chat = client.chats.create(
    model="gemini-3.5-flash"
)

while True:
    user_input = input("You: ")

    if user_input.lower() == "exit":
        break

    response = chat.send_message(
        message=user_input
    )

    print("AI:", response.text)
````
